```python
from dotenv import load_dotenv
import os
import requests
load_dotenv()
j_search_api = os.getenv("X-API-Key")
```

```python

headers = {
  "X-API-Key": j_search_api
}

response = requests.request(
    'GET',
    'https://api.openwebninja.com/jsearch/search-v2',
    params={
      "query":"AI engineer jobs in  banglore",
        "job_title": "AI Engineer",
        "location": "Bangalore",
        "location_type": "ANY",
        "years_of_experience": 3
      },
    headers=headers
)
data = response.json()      

jobs = data["data"]["jobs"]

print(jobs)
```
