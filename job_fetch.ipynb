```python
from dotenv import load_dotenv
import os

load_dotenv()
j_search_api = os.getenv("X-API-Key")
```

```python
import requests

headers = {
  "X-API-Key": j_search_api
}

response = requests.request(
    'GET',
    'https://api.openwebninja.com/jsearch/search-v2',
    params={"query":"AI engineer jobs in India"},
    headers=headers
)
data = response.json()      

jobs = data["data"]["jobs"]

print(jobs)
```
